```python
import pandas as pd
import numpy as np

from pathlib import Path

from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import BernoulliNB
from sklearn.metrics import accuracy_score
```

```python
DATA_DIR = Path("../data/CUB_200_2011/CUB_200_2011")

print(DATA_DIR.exists())
```

```python
attribute_labels_file = DATA_DIR / "attributes" / "image_attribute_labels.txt"

records = []

with open(attribute_labels_file, "r", encoding="utf-8") as f:
    for line in f:
        parts = line.split()

        # We only need:
        # image_id, attribute_id, is_present
        if len(parts) >= 3:
            image_id = int(parts[0])
            attribute_id = int(parts[1])
            is_present = int(parts[2])

            records.append((image_id, attribute_id, is_present))

attribute_labels = pd.DataFrame(
    records,
    columns=["image_id", "attribute_id", "is_present"]
)

print(attribute_labels.head())
print("Shape:", attribute_labels.shape)
```

```python
X_attributes = attribute_labels.pivot(
    index="image_id",
    columns="attribute_id",
    values="is_present"
)

X_attributes.columns.name = None
X_attributes = X_attributes.sort_index(axis=1)

print("Shape:", X_attributes.shape)
print(X_attributes.head())
```

```python
labels_file = DATA_DIR / "image_class_labels.txt"

image_labels = pd.read_csv(
    labels_file,
    sep=r"\s+",
    header=None,
    names=["image_id", "class_id"]
)

print(image_labels.head())
print("Shape:", image_labels.shape)
print("Unique classes:", image_labels["class_id"].nunique())
```

```python
data = X_attributes.reset_index().merge(
    image_labels,
    on="image_id"
)

print("Shape:", data.shape)
print(data.head())
```

```python
X = data.drop(columns=["image_id", "class_id"])
y = data["class_id"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
nb_model = BernoulliNB()

nb_model.fit(X_train, y_train)

y_pred = nb_model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Naive Bayes Accuracy:", accuracy)
print("Naive Bayes Accuracy (%):", accuracy * 100)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

knn_model = KNeighborsClassifier(n_neighbors=4)

knn_model.fit(X_train, y_train)

y_pred_knn = knn_model.predict(X_test)


knn_accuracy = accuracy_score(y_test, y_pred_knn)



print("KNN Accuracy:", knn_accuracy)
print("KNN Accuracy (%):", knn_accuracy * 100)

```

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score

svm_model = LinearSVC(
    C=1.0,
    random_state=42,
    max_iter=5000
)

svm_model.fit(X_train, y_train)

y_pred_svm = svm_model.predict(X_test)

svm_accuracy = accuracy_score(y_test, y_pred_svm)

print("SVM Accuracy:", svm_accuracy)
print("SVM Accuracy (%):", svm_accuracy * 100)
```

```python
from sklearn.multiclass import OneVsRestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

ovr_lr_model = OneVsRestClassifier(
    LogisticRegression(
        max_iter=2000,
        random_state=42
        
    )
)

ovr_lr_model.fit(X_train, y_train)

y_pred_ovr_lr = ovr_lr_model.predict(X_test)

ovr_lr_accuracy = accuracy_score(y_test, y_pred_ovr_lr)

print("One-vs-Rest Logistic Regression Accuracy:",
      ovr_lr_accuracy * 100, "%")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

pca_lr_model = Pipeline([
    ("scaler", StandardScaler()),
    ("pca", PCA(n_components=0.95)),
    ("lr", OneVsRestClassifier(
        LogisticRegression(
            max_iter=5000,
            # random_state=42
        )
    ))
])

pca_lr_model.fit(X_train, y_train)

y_pred_pca_lr = pca_lr_model.predict(X_test)

pca_lr_accuracy = accuracy_score(
    y_test,
    y_pred_pca_lr
)

n_components = pca_lr_model.named_steps["pca"].n_components_

print("PCA components retained:", n_components)
print(
    "Explained variance:",
    pca_lr_model.named_steps["pca"].explained_variance_ratio_.sum()
)
print(
    "One-vs-Rest Logistic Regression + PCA Accuracy:",
    pca_lr_accuracy * 100,
    "%"
)
```

```python
from sklearn.feature_selection import SelectKBest, f_classif

fs_lr_model = Pipeline([
    ("selection", SelectKBest(
        score_func=f_classif,
        k=210
    )),
    ("lr", OneVsRestClassifier(
        LogisticRegression(
            max_iter=2000,
            random_state=42
        )
    ))
])

fs_lr_model.fit(X_train, y_train)

y_pred_fs_lr = fs_lr_model.predict(X_test)

fs_lr_accuracy = accuracy_score(
    y_test,
    y_pred_fs_lr
)

print(
    "Selected features:",
    fs_lr_model.named_steps["selection"].get_support().sum()
)

print(
    "Feature Selection + One-vs-Rest Logistic Regression Accuracy:",
    fs_lr_accuracy * 100,
    "%"
)
```

```python
pca_fs_lr_model = Pipeline([
    ("scaler", StandardScaler()),

    ("pca", PCA(n_components=0.95)),

    ("selection", SelectKBest(
        score_func=f_classif,
        k=150
    )),

    ("lr", OneVsRestClassifier(
        LogisticRegression(
            max_iter=2000,
            random_state=42
        )
    ))
])

pca_fs_lr_model.fit(X_train, y_train)

y_pred_pca_fs_lr = pca_fs_lr_model.predict(X_test)

pca_fs_lr_accuracy = accuracy_score(
    y_test,
    y_pred_pca_fs_lr
)

pca_components = pca_fs_lr_model.named_steps["pca"].n_components_

selected_features = (
    pca_fs_lr_model
    .named_steps["selection"]
    .get_support()
    .sum()
)

explained_variance = (
    pca_fs_lr_model
    .named_steps["pca"]
    .explained_variance_ratio_
    .sum()
)

print("PCA components retained:", pca_components)
print("Explained variance:", explained_variance)
print("Selected PCA features:", selected_features)

print(
    "PCA + Feature Selection + One-vs-Rest Logistic Regression Accuracy:",
    pca_fs_lr_accuracy * 100,
    "%"
)
```
